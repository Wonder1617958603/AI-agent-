## 1 VSCode 基本介绍

vscode 是由微软推出的一款免费，开源且跨平台的代码编辑器，我们可以通过他的官网：https：//code.visualstudio.com/ 下载安装。以及阅读大量相关的帮助文档。因为VSCode是开源的，所以我们可以自由的访问它在github上的仓库，地址为：https://github.com/microsoft/vscode 。通过查看源代码根目录下的package.json文件，我们可以发现，VSCode其实是基于Electron构建的，Electron是一个使用Web技术构建跨平台桌面应用程序的框架。

vscode这款软件的组成，其实是里面嵌入了一个Chrome浏览器外加一个node.js运行时，整个VSCode我们能够看到的界面，其实都是由HTML，CSS和JavaScript构建的，并使用chrome来渲染出来的页面，如果你不相信，你可以点击VSCode顶部菜单栏的最后一项help，然后选择倒数第二项：Toggle developer tools，就可以看到弹出的Chrome开发者工具了。

VSCode近些年越来越受开发者欢迎，是因为它的身上存在这么几个亮点：
1. 第一是打开速度非常快，用VSCode来打开一个项目，基本上可以说是秒开，即使是很大的项目，也可以非常快的被打开。

2. 第二是它支持的编程语言越来越多，几乎支持所有主流的编程语言，包括Python，Java，C++，JavaScript，TypeScript，PHP，Go，Ruby，Swift，Kotlin等。只要你安装了对应的插件，就可以直接在VSCode中编写和运行代码。

3.第三是拥有丰富的扩展，VSCode的扩展非常丰富，首先从量上来看，VSCode的扩展数量已经上万，涵盖了方方面面，你能想到的，几乎都有对应的扩展。

4.第四是涵盖应用程序的整个生命周期，从代码的编写，到调试，再到发布，VSCode都可以提供对应的支持。

### 1.1 visual studio code和visual studio的区别
接下来，我们来看两个可能会遇到的并且有点令人迷惑的英文词汇：visual studio code和visual studio。其实vscode的全称是visual studio codex，作为微软于2015年推出的一款代码编辑器，它还有一个老大哥，那就是visual studio。

visual studio是微软于1997年推出的一款功能强大的IDE，但它主要用于微软自家的一些技术开发，比如说.net core开发。VSCode作为一款新产品，在创建之初，就从他大哥身上借鉴了很多很好的地方，比如说后面课程会学习的workspace概念，如果你使用过visual studio，你会发现这和visual studio的solution很类似，同时VSCode也摒弃了他大哥身上一些不太好的地方，最大改善就是让整个软件变得更加轻量。VSCode安装过后大概占用不到300MB的硬盘空间，而visual studio呢？没有2-3个g根本搞不定。

除了有老大哥visual studio，VSCode他还有一位双胞胎兄弟，叫做insiders editon，这个版本是微软专门为开发者准备的，它拥有最新的功能，但同时它也是最不稳定的，所以如果你想要使用最新的功能，但又不想遇到太多bug，那么insiders edition绝对是一个不错的选择。

### 1.2 界面布局与功能介绍
在第一章中我们了解了VSCode的基本发展历程。

#### 1.2.1 界面布局
VSCode的用户界面一共由五大部分组成，分别是：
* 底部的Status Bar（状态栏），用于显示当前被打开文件的一些信息。
* 最左侧的Activity Bar（活动栏），用于显示当前打开的文件夹，文件，以及一些其他功能。
* 中间的Editor（编辑器），用于显示当前被打开的文件内容。
* 编辑器下方的panels（面板），它包含4个不同的面板，其中terminal面板我们会经常用到。

![VSCode界面布局](../attachments/vscode-layout.png)

#### 1.2.2 资源管理器（explorer）
Explorer面板是VSCode中一个非常重要的面板，它可以帮助我们管理当前打开的文件夹，文件，以及一些其他功能。我们可以看到，整个功能区被分为了3部分，从上到下分别是：
* open editors（打开的编辑器），它显示了当前被打开的文件。
* 项目本身
* outliner（大纲），它显示了当前被打开文件的目录结构。
* timeline（时间轴），它显示了当前被打开文件的修改历史。

![explorer](../attachments/VSCode-explore.png)

在open editors里面，它以列表的形式列出了当前所有被打开的文件，这里面包含了一个非常好用的功能，我们可以将鼠标移动到open editors的右侧，这时会出现三个功能键，通过点击最右侧那个带有叉叉图标的按键，我们就可以一键关闭当前打开的文件。在实际的项目开发过程中，随着时间的推移，我们打开的文件越来越多，这时候我常常希望关闭所有被打开的文件，来让整个编辑器看起来更加清爽，而利用刚刚说的那个带有叉叉图标的按键，我们就很方便的完成这种操作。

最后，我们看看最下方的outline，outline我用自己的话总结它的作用就是：换个角度看代码。对于一份源代码，outline中列出的中列出的视图常常会以另外一个角度来看待，比如当前源代码是Html,outline中视图就会以dom树的结构来看待。再比如当前源代码是面向对象编程语言，它就会从类，方法，属性等角度来看待。

这样从不同角度来看待源代码会带来什么好处呢？好处有两点：
* 第一，当代码过长时，通过outline中列出视图信息，我们可以很容易理解代码的组织架构；
* 第二，当我们想要修改某一个功能时，通过outline中列出的视图信息，我们可以很容易的找到需要修改的代码位置。


#### 1.2.3 搜索工具
点击Search，我们可以在Slide Bar顶部看到两个输入框，第一个输入框是搜索框，第二个输入框是替换框。相较于替换，在平时工作中使用搜索会更多一些。通过搜索，我们可以很方便的找到某个单词或者某个句子在整个项目中出现过的地方，通过查看搜索词在其他文件中是如何被使用，如果是跟踪一个bug，那么就能通过这种方式来快速追踪bug的来源，而如果我们是刚接手一个项目，就很有可能对其中一些函数或者变量感到陌生，这时候就可以通过搜索来快速了解这些函数或者变量的作用。

搜索和替换的使用方法很简单，直接在输入框中输入相应的单词就可以，但是VSCode还针对搜索和替换提供了一些高级选项，通过这些高级选项，我们可以进一步缩小查询结果的范围，但遇到比较大的项目时，开启这些选项还是很有用的，可以帮助我节省查看搜索结果所花费的时间。

#### 1.2.4 源代码管理（source control）
通过source control，我们可以很方便在VSCode里完成对Git的相关操作。不过，为了使用source control，我们必须满足两个条件：
* 第一，当前项目必须已经初始化过Git仓库，如果项目还没有初始化过Git仓库，我们可以通过VSCode的source control面板来完成初始化。
* 第二，当前项目必须已经安装了Git，如果项目还没有安装Git，我们可以通过VSCode的source control面板来完成安装。

Git在项目开发中扮演着非常重要的角色，它可以帮助我们管理代码的版本，同时也可以帮助我们管理代码的提交，回退，以及合并等操作。

在搭建Git环境之前，我们先来看一下Git命令。

* git init：初始化Git仓库
* git add：添加文件到Git仓库
* git commit：提交文件到Git仓库
* git push：推送文件到Git仓库
* git pull：拉取文件到Git仓库 

为了安全，边界的完成与远程仓库的通讯，我们常常使用SSH来完成与远程仓库的通讯。而为了使用ssh，我们需要在VSCode中配置SSH密钥。所以整个Git环境的搭建有很大一部分精力是花在ssh的配置上。

![source control](../attachments/Source-control.png)

##### （1）安装Git
首先需要前往Git官网下载Git的安装包，下载地址为：https://git-scm.com/downloads 。下载完成后，直接运行安装包，按照安装向导的提示完成安装。

##### （2）完成Git的基础配置
安装完Git后，我们需要完成Git的两个基础配置：user.name和user.email，这样我们就能知道每次commit的作者是谁，他的联系邮箱又是什么，为了配置这两项，需要输入以下命令：

```python
Git config --global user.name "your name"
Git config --global user.email "your email"
```

##### （3）生成ssh公钥私钥
从这一步开始，我们就要使用命令行来完成ssh的配置了。


##### （4）将ssh公钥添加到github
因为后面我们使用github来完成远程仓库的创建，所以需要将ssh公钥添加到github上。完成这个配置很简单，只需要在github上点击右上角的头像，然后选择Settings，在Settings页面中点击SSH and GPG keys，然后点击New SSH key，在弹出的对话框中输入ssh公钥，然后点击Add SSH key即可。这样，以后我们本地电脑上的私钥与github上的公钥就能自动匹配，从而完成ssh的配置，ssh通信也就完成了。

##### （5）测试ssh连接
完成ssh的配置后，我们就可以测试ssh连接了。在VSCode中，我们可以通过点击VSCode顶部的Terminal，然后选择New Terminal来打开一个新的终端，然后输入以下命令：

```python
ssh -T git@github.com
```

如果连接成功，就会看到类似如下的输出：

#### 1.2.5 在vscode中使用git
在VSCode中使用Git有两种方式，一种是使用VSCode自带的source control面板，另一种是使用VSCode的terminal面板。

这两种方式结果完全一致，但两份都需要掌握。因为：
* 第一，可以帮助复习Git命令，因为在很多时候我们只能通过命令行来完成Git操作；比如在远程的Linux服务器上，我们只能通过命令行来完成Git操作。所以，我们可以使用Git的GUI工具（比如Sorurcetree，GitHubdesktop，gitkraken以及我们即将学习的source control），但是一定不要忘记命令行操作。
* 第二，可以帮助我们理解Git的原理，因为Git的原理非常复杂，通过命令行操作，可以帮助我们更好的理解Git的原理。
* 第三，采用这种方式后，我们可以很好的进行比较：哪些操作使用source control面板，哪些操作使用terminal面板。

git操作有很多，在这一节课里，我们学习利用本地环境就能完成操作的Git命令，他们分别是：
* git init：初始化Git仓库
* git add：添加文件到Git仓库
* git commit：提交文件到Git仓库
* git branch：创建分支
* git checkout：切换分支
* git merge：合并分支
* git status：查看仓库状态
* git log：查看提交历史
* git reset：回退版本

然后是与远程仓库相关的命令：
* git push：推送文件到Git仓库
* git pull：拉取文件到Git仓库 
* git clone：克隆远程仓库

#### 1.2.6 调试工具（debug）
Debug也就是调试，在软件项目开发过程中调试扮演着非常重要的角色。当我们编写的程序遇到问题时，我们可以利用调试来快速定位问题发生的原因；当我们对一段代码逻辑不是很清楚时，我们也可以利用调试来帮助我们理解代码内部的逻辑，可以看到调试功能对开发人员来说有多么重要。

VS Code中涉及调试的知识非常多，我们将知识分为两类，一类叫“通用知识”，这些通用知识，无论我们使用VSCode来调试什么项目，它都可以用上，甚至是几十我们不用VScode转而使用其他编辑器。还有一类，叫“和具体编程语言相关的调试知识”，我们知道，VSCode支持很多编程语言的调试，而不同的编程语言在调试时，需要使用不同的调试配置。

不过为了学习调试的“通用知识”，我们还是必须要选择一门编程语言来写调试的演示项目，这里选择基于Node.js的JavaScript语言。为什么会是它呢？回答这个问题之前，我们先看一下，如果想在VSCode中调试一个项目，需要完成哪些配置。
* 搭建相应编程语言的开发环境：比如说，我们想调试Java程序，那我们就必须先要在本地搭建好Java的开发环境，保证可以在命令行中调用Java和javac命令。
* 安装相应编程语言用于调试的扩展：不同的编程语言在调试时，还需要借助专门的扩展。比如说，我们想调试PHP程序，那我们就必须安装一个叫PHP Debug的扩展。
* 配置launch.json文件：这个文件是调试的核心，它定义了调试的配置，包括调试器类型，调试器路径，调试器参数等。

可以看到，如果我们想要在VScode中学习调试，要准备的东西还是挺多的，不过，如果我们使用的是Node.js的话，我们可以省略其中的条件2和条件3，这是因为VSCode默认只支持Node.js的调试（P.S.在第三课中，我们有提到VSCode的内部其实是嵌入了一个Chrome浏览器外加一个Node.js运行时，这或许就是VSCode和Node.js这么亲近的原因吧）。所以，对于调试的”通用知识“，我们将使用Node.js来完成演示。
